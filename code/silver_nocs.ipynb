```python
df = spark.read.format('csv')\
              .option('header',True)\
              .option('inferSchema',True)\
              .load('abfss://bronze@xc797demo.dfs.core.windows.net/nocs')
```

```python
display(df)
```

```python
df = df.drop('country')
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df.withColumn('tag', split(col('tag'), '-')[0]).display()
```

```python
df_clean = df.withColumn('tag_original', col('tag')) \
             .withColumn('tag', split(col('tag'), '-')[0])

df_clean.display()
```

```python
# create ext table in delta lake
df.write.format('delta')\
.mode('append')\
.option("path", 'abfss://silver@xc797demo.dfs.core.windows.net/nocs')\
    .saveAsTable("olympic.silver.nocs")
```
